```python
import pandas as pd

df = pd.read_csv('../data/raw/data.csv')

df.head(30)
```

```python
print(df.shape)
print(df.columns)
```

```python
print(df.isna().sum())
```

```python
print(df["text"].duplicated().sum())
```

```python
df = df.dropna(subset=["text"])
print(df.shape)
```

```python
df = df.drop_duplicates(subset=["text"])
print(df.shape)
```

```python
df["text"] = df["text"].str.strip().str.replace(r"\s+", " ", regex=True)
df.head(10)
```

```python
df = df[df["text"].str.len() > 50]
df.head(15)
```
