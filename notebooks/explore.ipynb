```python
import pandas as pd

df = pd.read_csv('..\\data\\raw\\data.csv')

print(df.shape)
print(df.columns)

df.head()
```

```python
print("Пропуски:\n", df.isna().sum())
```

```python
print(df["text"].duplicated().sum())
```

```python
df = df.dropna(subset=["text"])
print(df.shape)
```

```python
df = df.drop_duplicates(subset=["text"])
print(df.shape)
```

```python
df["text"] = df["text"].str.strip().str.replace(r"\s+", " ", regex=True)
df.head(10)
```
